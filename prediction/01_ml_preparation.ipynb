```python
df = pd.read_csv(DATASET_PATH, low_memory=False)

print("Dimensions :", df.shape)
display(df.head())
```

```python
selected_cols = [
    "sexe", "catu", "catv", "atm", "lum",
    "col", "choc", "manv", "plan", "surf",
    "nbv", "secu1", "infra", "age", "place"
]

target_col = "grav"

required_cols = [
    col for col in selected_cols if col != "age"
] + ["grav", "an", "an_nais", "Num_Acc"]

missing_cols = [
    col for col in required_cols
    if col not in df.columns
]

print("Colonnes manquantes :", missing_cols)

if missing_cols:
    raise ValueError(f"Colonnes absentes : {missing_cols}")
else:
    print("Toutes les colonnes nécessaires sont présentes.")
```

```python
missing_stats = df[required_cols].isna().sum().to_frame("Valeurs manquantes")

missing_stats["Pourcentage (%)"] = (
    missing_stats["Valeurs manquantes"] / len(df) * 100
).round(2)

display(
    missing_stats.sort_values(
        "Pourcentage (%)",
        ascending=False
    )
)
```

```python
# Créer un dataset indépendant
df_ml = df[required_cols].copy()

# Convertir les années en valeurs numériques
df_ml["an"] = pd.to_numeric(df_ml["an"], errors="coerce")
df_ml["an_nais"] = pd.to_numeric(
    df_ml["an_nais"],
    errors="coerce"
)

# Calculer l'âge au moment de l'accident
df_ml["age"] = df_ml["an"] - df_ml["an_nais"]

# Identifier les âges incohérents
invalid_age = (
    df_ml["age"].notna()
    & ~df_ml["age"].between(0, 110)
)

print("Âges incohérents :", invalid_age.sum())

# Remplacer les âges incohérents par NaN
df_ml.loc[invalid_age, "age"] = np.nan

# Vérifier les résultats
print("Dimensions :", df_ml.shape)
print("Âges manquants :", df_ml["age"].isna().sum())

display(
    df_ml[["an", "an_nais", "age"]].head(10)
)
```

```python
categorical_cols = [
    "sexe", "catu", "catv", "atm", "lum",
    "col", "choc", "manv", "plan", "surf",
    "secu1", "infra", "place"
]

for col in categorical_cols:
    print(f"\n{'=' * 40}")
    print(f"{col.upper()} : {df_ml[col].nunique()} catégories")
    print(df_ml[col].value_counts(dropna=False).to_string())
```

```python
numeric_cols = ["age", "nbv"]

for col in numeric_cols:
    values = pd.to_numeric(df_ml[col], errors="coerce")

    print(f"\n{col.upper()}")
    print(values.describe())
    print("Valeurs non numériques :", df_ml[col].notna().sum() - values.notna().sum())
```

```python
# Vérifier les valeurs particulières
for col in ["sexe", "nbv", "place", "plan", "secu1"]:
    print(f"\n{'=' * 45}")
    print(f"COLONNE : {col.upper()}")
    print(f"{'=' * 45}")

    print(
        df_ml[col]
        .value_counts(dropna=False)
        .to_string()
    )
```

```python
nbv_numeric = pd.to_numeric(df_ml["nbv"], errors="coerce")

invalid_nbv = df_ml[
    df_ml["nbv"].notna() & nbv_numeric.isna()
]

print("Nombre de valeurs non numériques :", len(invalid_nbv))

print("\nValeurs concernées :")
display(invalid_nbv["nbv"].value_counts(dropna=False))
```

```python
# Nettoyer le nombre de voies
df_ml["nbv"] = pd.to_numeric(
    df_ml["nbv"].astype("string").str.strip(),
    errors="coerce"
)

# -1 signifie une information inconnue
df_ml.loc[df_ml["nbv"] == -1, "nbv"] = np.nan

# Vérifier les résultats
print("Valeurs manquantes nbv :", df_ml["nbv"].isna().sum())
print("Valeurs manquantes age :", df_ml["age"].isna().sum())

display(df_ml[["nbv", "age"]].describe())
```

```python
categorical_cols = [
    "sexe", "catu", "catv", "atm", "lum",
    "col", "choc", "manv", "plan", "surf",
    "secu1", "infra", "place"
]

for col in categorical_cols:
    df_ml[col] = df_ml[col].astype("string").str.strip()

    df_ml[col] = df_ml[col].replace({
        "-1": "Non renseigné",
        "#ERREUR": "Non renseigné",
        "#VALEURMULTI": "Non renseigné"
    })

print("Nettoyage des catégories terminé.")

print("\nValeurs de sexe :")
print(df_ml["sexe"].value_counts(dropna=False))

print("\nValeurs de place :")
print(df_ml["place"].value_counts(dropna=False))
```

```python
# Variables utilisées par le formulaire
features = [
    "sexe", "catu", "catv", "atm", "lum",
    "col", "choc", "manv", "plan", "surf",
    "nbv", "secu1", "infra", "age", "place"
]

# Variable cible
target = "grav"

# Identifiant conservé pour séparer les accidents
group_col = "Num_Acc"

# Dataset final pour le Machine Learning
df_ml_final = df_ml[
    [group_col] + features + [target]
].copy()

# Contrôles
assert len(df_ml_final) == len(df_ml)
assert df_ml_final.columns.is_unique
assert df_ml_final["grav"].notna().all()

print("Dimensions :", df_ml_final.shape)
print("Accidents uniques :", df_ml_final["Num_Acc"].nunique())

print("\nRépartition des catégories de gravité :")
print(df_ml_final["grav"].value_counts(dropna=False))

display(df_ml_final.head())

# Export
df_ml_final.to_csv(
    OUTPUT_PATH,
    index=False,
    encoding="utf-8"
)

print("\nDataset ML enregistré :", OUTPUT_PATH)
```

```python
valid_gravities = [
    "Indemne",
    "Blessé léger",
    "Blessé hospitalisé",
    "Tué"
]

# Conserver uniquement les gravités connues
df_ml_final = df_ml_final[
    df_ml_final["grav"].isin(valid_gravities)
].copy()

print("Dimensions après nettoyage :", df_ml_final.shape)
print("Accidents uniques :", df_ml_final["Num_Acc"].nunique())

print("\nRépartition finale des gravités :")
print(df_ml_final["grav"].value_counts())

# Réexporter le dataset nettoyé
df_ml_final.to_csv(
    OUTPUT_PATH,
    index=False,
    encoding="utf-8"
)

print("\nDataset ML mis à jour :", OUTPUT_PATH)
```
